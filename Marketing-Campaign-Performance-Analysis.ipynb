## Data Visualization

```python
import matplotlib.pyplot as plt
def income_impact():
    df['Total_Spending'] = df[['MntWines', 'MntFruits', 'MntMeatProducts', 
                               'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']].sum(axis=1)
   
    plt.scatter(df['Income'], df['Total_Spending'])
    plt.xlabel('Income')
    plt.ylabel('Total Spending')
    plt.show()

income_impact()
```

```python

def coustomer_loyalty():
    loyalty_data = df.groupby('Response')[['Total_Spending', 'Recency', 'NumWebPurchases', 'NumStorePurchases']].mean()

    
    loyalty_data['Total_Spending'].plot(kind='bar', title='Total Spending by Campaign Response')
    plt.xlabel('Campaign Response')
    plt.ylabel('Average Total Spending')
    plt.show()

    loyalty_data['Recency'].plot(kind='bar', title='Recency by Campaign Response')
    plt.xlabel('Campaign Response')
    plt.ylabel('Average Recency')
    plt.show()

    loyalty_data['NumWebPurchases'].plot(kind='bar', title='Web Purchases by Campaign Response')
    plt.xlabel('Campaign Response')
    plt.ylabel('Average Web Purchases')
    plt.show()

    loyalty_data['NumStorePurchases'].plot(kind='bar', title='Store Purchases by Campaign Response')
    plt.xlabel('Campaign Response')
    plt.ylabel('Average Store Purchases')
    plt.show()


coustomer_loyalty()
```

```python


def trends():
    df['Month'] = pd.to_datetime(df['Dt_Customer'], errors='coerce').dt.month


    monthly_spending = df.groupby('Month')['Total_Spending'].mean()


    monthly_spending.plot()
    plt.xlabel('Month')
    plt.ylabel('Average Total Spending')
    plt.show() 

trends()
```

```python

def spending_habits():
    family_data = df.groupby(['Marital_Status', 'Kidhome'])['Total_Spending'].mean()

    family_data.unstack().plot(kind='bar')
    plt.xlabel('Marital Status')
    plt.ylabel('Average Total Spending')
    plt.show()

spending_habits()
```

```python
def display_salary(user_id):
    user_data = df[df['ID'] == user_id]  
    if not user_data.empty:
        salary = user_data['Income'].values[0]  
        print(f"The salary for User ID {user_id} is: {salary}")
    else:
        print("User ID not found.")

user_id = int(input("Enter your User ID to get income: "))
display_salary(user_id)
```
